```python
import numpy as np; import matplotlib.pyplot as plt

# dx/dt=r*t; exponentional growth
r = 0.34
x0 = 2 # starting x value
t0 = 0 # starting t value
t_max = 10 # simulation limit
dt = 0.33 # step

t = np.arange(t0, t_max + dt, dt)
n_steps = len(t)

x = np.zeros(n_steps)
x[0] = x0

for i in range(n_steps - 1) :
    dxdt = r*x[i]

    x[i+1] = x[i] + dxdt * dt # x_(i+1) = x_i + f(x_i, t_i) * dt

plt.plot(t, x, marker = 'o', markersize = 3, label = 'Exponential Growth')
plt.xlabel('Time (t)')
plt.ylabel('Value (x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np; import matplotlib.pyplot as plt

# dT/dt=-k(T-A)
k = 0.1 # cooling coefficient
Temp0 = 0 # starting temperature; = T_0
A = 10 # surrounding temperature
t0 = 0 # starting t value
t_max = 50 # simulation limit
dt = 1 # step

t = np.arange(t0, t_max + dt, dt)
n_steps = len(t)

Temp = np.zeros(n_steps)
Temp[0] = Temp0

for i in range(n_steps - 1) :
    dTempdt = (-k) * (Temp[i] - A)

    Temp[i+1] = Temp[i] + dTempdt * dt # Temp_(i+1) = Temp_i + f(Temp_i, t_i) * dt

plt.plot(t, Temp, marker = 'o', markersize = 3, label = "Newton's Law")
plt.xlabel('Time (t)')
plt.ylabel('Temperature (T)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np; import matplotlib.pyplot as plt

# dx/dt = \sigma * (y-x)
# dy/dt = x * (\rho - z) - y
# dz/dt = xy - \beta * z
# we use the "Lorenz system"
sigma = 10
rho = 28
beta = 8/3
x0 = 0.2
y0 = 0.3
z0 = 0.4
t0 = 0 # starting t value
t_max = 50 # simulation limit
dt = 0.007 # step

t = np.arange(t0, t_max + dt, dt)
n_steps = len(t)

x = np.zeros(n_steps)
x[0] = x0

y = np.zeros(n_steps)
y[0] = y0

z = np.zeros(n_steps)
z[0] = z0

for i in range(n_steps - 1) :
    dxdt = sigma * (y[i] - x[i])
    dydt = x[i] * (rho - z[i]) - y[i]
    dzdt = x[i] * y[i] - beta * z[i]


    x[i+1] = x[i] + dxdt * dt # Temp_(i+1) = Temp_i + f(Temp_i, t_i) * dt
    y[i+1] = y[i] + dydt * dt
    z[i+1] = z[i] + dzdt * dt

plt.plot(x, z, lw=0.5, color='purple')
plt.xlabel('X')
plt.ylabel('Z')
plt.title('Lorenz Attractor: X-Z Projection')
plt.grid(True)
plt.show()
```
